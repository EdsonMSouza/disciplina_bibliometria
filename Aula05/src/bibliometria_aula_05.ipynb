### Carga de Dados e Exploração Inicial

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração de estilo
sns.set(style="whitegrid")

# Carregar dados (Exemplo)
df = pd.read_csv('dados_bibliometricos.csv', sep=";")

# Inspeção
df.head()
```

### Visualizando a Produção (Barplot)

```python
# Contagem por ano
publicacoes_por_ano = df['Year'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
sns.barplot(x=publicacoes_por_ano.index, y=publicacoes_por_ano.values, color='skyblue')

plt.title('Evolução da Produção Científica', fontsize=14)
plt.xlabel('Ano')
plt.ylabel('Número de Artigos')
plt.xticks(rotation=45)
plt.show()
```

### Distribuição de Citações (Histograma & Boxplot)

```python
plt.figure(figsize=(12, 5))

# Histograma
plt.subplot(1, 2, 1)
sns.histplot(df['Citations'], bins=30, kde=True)
plt.title('Distribuição de Citações')

# Boxplot (Para ver outliers)
plt.subplot(1, 2, 2)
sns.boxplot(x=df['Citations'])
plt.title('Boxplot de Citações')

plt.show()
```

### Correlação e Colaboração (Scatterplot)

```python
# Criar coluna de número de autores (se a coluna Authors for strings separadas por ;)
df['Num_Authors'] = df['Authors'].apply(lambda x: len(str(x).split(';')))

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Num_Authors', y='Citations', alpha=0.6)

plt.title('Relação: Tamanho da Equipe vs. Impacto')
plt.xlabel('Número de Coautores')
plt.ylabel('Citações Recebidas')
plt.show()
```

### Calcular o H-index de um autor

```python
import pandas as pd

# 1. Carregar os dados
df = pd.read_csv('dados_bibliometricos.csv', sep=';')

# 2. Definir a função lógica do H-Index
def calcular_h_index(lista_citacoes):
    """
    Calcula o índice h dado uma lista de números de citações.
    Lógica: Ordenar (decrescente) e encontrar o último ponto onde Citações >= Rank.
    """
    # Ordenar do maior para o menor
    citacoes_ordenadas = sorted(lista_citacoes, reverse=True)

    h = 0
    # Enumerate nos dá o índice (começando em 0) e o valor
    for i, citacoes in enumerate(citacoes_ordenadas):
        rank = i + 1  # O rank é a posição (1º, 2º, 3º...)

        if citacoes >= rank:
            h = rank
        else:
            # Se a citação for menor que o rank, paramos a contagem
            break

    return h

# 3. Definir o autor alvo e filtrar o DataFrame
autor_alvo = "Silva, A.B."

# Filtramos as linhas onde a coluna 'Authors' contém o texto "Silva, A.B."
# O parâmetro regex=False trata a string literalmente (evita erros com caracteres especiais)
df_autor = df[df['Authors'].str.contains(autor_alvo, na=False, regex=False)]

# 4. Aplicar a função e exibir resultados
h_index = calcular_h_index(df_autor['Citations'])

print(f"--- Relatório para {autor_alvo} ---")
print(f"Total de Artigos: {len(df_autor)}")
print(f"Lista de Citações: {sorted(df_autor['Citations'].tolist(), reverse=True)}")
print(f"Índice H calculado: {h_index}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados do autor Silva, A.B.
citacoes = [95, 45, 41, 32, 28, 10, 5]
ranks = np.arange(1, len(citacoes) + 1) # Cria [1, 2, 3, 4, 5, 6, 7]

# Configuração do Gráfico
plt.figure(figsize=(10, 6))

# 1. Plotar as Barras de Citação
plt.bar(ranks, citacoes, color='#a1c9f4', label='Citações do Artigo', alpha=0.7)

# 2. Plotar a Linha de Referência (y = x)
# Esta linha representa a "Barreira do H": onde Rank == Citações
limite = max(ranks) + 1
plt.plot([0, limite], [0, limite], color='red', linestyle='--', linewidth=2, label='Linha de Corte (Rank = Citações)')

# 3. Marcar o Ponto do H-Index (H=6)
h_index = 6
plt.scatter([h_index], [citacoes[h_index-1]], color='green', s=100, zorder=5, label=f'H-Index ({h_index})')

# Destacar a área válida (H-Core)
plt.fill_between([0, h_index], [0, 0], [h_index, h_index], color='green', alpha=0.1)
plt.vlines(h_index, 0, h_index, colors='green', linestyles=':')
plt.hlines(h_index, 0, h_index, colors='green', linestyles=':')

# 4. Anotações para a Aula
# Mostrar o sucesso no 6
plt.annotate(f'Rank 6\n10 Citações\n(10 > 6)',
             xy=(6, 10), xytext=(4.5, 20),
             arrowprops=dict(facecolor='green', shrink=0.05))

# Mostrar a falha no 7
plt.annotate(f'Rank 7\n5 Citações\n(5 < 7) -> FALHA',
             xy=(7, 5), xytext=(7.5, 15),
             arrowprops=dict(facecolor='red', shrink=0.05))

# Estilização
plt.title('Visualização Geométrica do Índice H (Silva, A.B.)', fontsize=14)
plt.xlabel('Ranking do Artigo (1 = Mais citado)', fontsize=12)
plt.ylabel('Número de Citações', fontsize=12)
plt.xticks(ranks) # Forçar mostrar todos os números no eixo X
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.show()
```
