**A Biblioteca Plotly**

Plotly é uma biblioteca de visualização de dados voltada para gráficos interativos. Ela existe em duas frentes principais: Plotly.js (JavaScript) e Plotly para Python (além de R e outras linguagens). No ecossistema Python, ela costuma ser usada para exploração de dados, dashboards e relatórios interativos em Jupyter, Colab e aplicações web.

O diferencial é a interatividade nativa. Em praticamente qualquer gráfico você tem zoom, pan, seleção de pontos, tooltips ao passar o mouse, ocultar e mostrar séries na legenda, exportar imagem e controlar a visualização sem precisar reprocessar o gráfico. Isso é especialmente útil quando você quer inspecionar outliers, comparar grupos e fazer leituras pontuais de dados.

No Python, há dois níveis de API. O Plotly Express (plotly.express) é uma camada de alto nível para criar gráficos rapidamente a partir de um DataFrame, com sintaxe curta e suporte direto a mapeamentos como x, y, cor, tamanho, facetas e rótulos. Já o plotly.graph_objects é o nível mais detalhado, em que você define “traces” e “layouts” explicitamente, útil quando precisa de controle fino de eixos, anotações, subcomponentes e composições complexas.

Plotly também se integra bem com Dash, que é o framework da própria Plotly para construir dashboards web em Python, com callbacks reativos, componentes de UI e atualização dinâmica de gráficos. Em cenários simples, dá para combinar Plotly com ipywidgets em notebooks, como no seu exemplo, criando painéis interativos sem sair do ambiente de desenvolvimento.

Em termos de saída, Plotly gera HTML e JSON por trás, o que facilita compartilhar visualizações como arquivos HTML, incorporar em páginas, ou publicar em ambientes que suportem conteúdo interativo. Para publicação científica, ele é útil em apêndices online, relatórios técnicos e exploração de dados, enquanto para artigos estáticos geralmente se exporta imagem em alta resolução ou se usa alternativas mais orientadas a figuras estáticas.

# Exemplo 1 - Análise Bibliométrica: Impacto vs. Temporalidade por Cluster

O código faz três coisas, em sequência: monta um pequeno conjunto de dados bibliométricos fictício, transforma esse conjunto em um DataFrame do pandas e, por fim, gera um gráfico de dispersão interativo no Plotly para visualizar padrões de impacto ao longo do tempo por área temática.

Primeiro, ele importa bibliotecas. O pandas é usado para estruturar os dados em formato tabular. O ```plotly.express``` é usado para criar o gráfico de forma direta.

Em seguida, ele define um dicionário chamado ```data``` com 12 registros (dois por ano, de 2018 a 2023). Cada registro tem: Ano de publicação, identificador do artigo, número de citações, um rótulo de cluster temático (IA, Educação, Saúde) e um valor de “Fator_Impacto” (um número que, no exemplo, representa algo como JCR ou CiteScore). Depois ele faz ```df = pd.DataFrame(data)```, ou seja, cria a tabela em memória.

Na parte do gráfico, ele cria um scatter plot com ```px.scatter(...)``` usando os seguintes mapeamentos:

* Eixo X (```x="Ano"```): ano do artigo.

* Eixo Y (```y="Fator_Impacto"```): fator de impacto associado ao artigo.

* Tamanho do ponto (```size="Citações"```): pontos maiores indicam mais citações.

* Cor do ponto (```color="Cluster"```): cada cor representa um cluster temático.

* Texto ao passar o mouse (```hover_name="Artigo"```): ao “hover”, aparece o nome do artigo.

* ```log_x=False```: confirma que o eixo X não será logarítmico (aqui não muda nada porque “Ano” já é linear e discreto).

* ```size_max=60```: limita o maior tamanho de bolha para não dominar o gráfico.

* ```title=...```: define o título do gráfico.

Depois, ```fig.update_layout(...)``` ajusta a apresentação:
* ```template="plotly_white"```: fundo branco e estilo mais “limpo”.
* ```xaxis_title, yaxis_title```: títulos dos eixos em português.
* legend_title: título da legenda.

Por fim, ```fig.show()``` abre o gráfico interativo (no Jupyter Notebook, no navegador ou no ambiente que estiver executando). O resultado é um gráfico onde você consegue ver, simultaneamente, temporalidade (X), “impacto” (Y), volume de citações (tamanho) e tema (cor), além de identificar cada ponto pelo nome do artigo ao passar o mouse.

````python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Criando um dataset fictício para demonstração bibliométrica
data = {
    'Ano': [2018, 2018, 2019, 2019, 2020, 2020, 2021, 2021, 2022, 2022, 2023, 2023],
    'Artigo': ['Artigo A', 'Artigo B', 'Artigo C', 'Artigo D', 'Artigo E', 'Artigo F',
               'Artigo G', 'Artigo H', 'Artigo I', 'Artigo J', 'Artigo K', 'Artigo L'],
    'Citações': [120, 45, 150, 80, 200, 30, 95, 300, 50, 180, 10, 5],
    'Cluster': ['IA', 'Educação', 'IA', 'Saúde', 'IA', 'Educação', 'Saúde', 'IA', 'Saúde', 'IA', 'Educação', 'Saúde'],
    'Fator_Impacto': [4.5, 1.2, 4.5, 3.0, 5.1, 1.2, 3.0, 6.2, 3.0, 5.5, 1.5, 2.8]
}
# Criação do DataFrame (conjunto de dados)
df = pd.DataFrame(data)

# Criação da figura
fig = px.scatter(df,
                 x="Ano",
                 y="Fator_Impacto",
                 size="Citações",
                 color="Cluster",
                 hover_name="Artigo",
                 log_x=False,
                 size_max=60,
                 title="Análise Bibliométrica: Impacto vs. Temporalidade por Cluster")

# Melhorando o layout para apresentação acadêmica
fig.update_layout(
    template="plotly_white",
    xaxis_title="Ano de Publicação",
    yaxis_title="Fator de Impacto (JCR/CiteScore)",
    legend_title="Área Temática"
)

fig.show()
````

# Exemplo 3 - Mini dashboard interativo

````python
from google.colab import output
output.enable_custom_widget_manager()

import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display

# Carregando o seu arquivo direto do Github
df = pd.read_csv('/content/dataset_exemplo.csv')

# Preparando os filtros
lista_editoras = ['Todas'] + sorted(df['Publisher'].dropna().unique().tolist())
min_ano = int(df['Year'].min())
max_ano = int(df['Year'].max())

# Criando os componentes do dashboard
seletor_editora = widgets.Dropdown(options=lista_editoras, value='Todas', description='Editora:')
seletor_ano = widgets.IntSlider(value=min_ano, min=min_ano, max=max_ano, description='Ano inicial:')
botao_atualizar = widgets.Button(description="Atualizar Painel", button_style='primary')
area_grafico = widgets.Output()

def atualizar(b):
    with area_grafico:
        area_grafico.clear_output()

        # Lógica de filtragem
        dados_filtrados = df[df['Year'] >= seletor_ano.value]
        if seletor_editora.value != 'Todas':
            dados_filtrados = dados_filtrados[dados_filtrados['Publisher'] == seletor_editora.value]

        if dados_filtrados.empty:
            print("Nenhum dado encontrado com esses filtros.")
            return

        # Criando o gráfico de volume por ano
        fig_volume = px.bar(
            dados_filtrados.groupby('Year').size().reset_index(name='Quantidade'),
            x='Year', y='Quantidade',
            title='Produção Científica no Período Selecionado',
            template='plotly_white'
        )

        # Criando o gráfico de citações
        fig_citacoes = px.scatter(
            dados_filtrados, x='Year', y='Cited by',
            size='Cited by', hover_name='Title',
            title='Impacto (Citações) por Artigo',
            template='plotly_white'
        )

        fig_volume.show()
        fig_citacoes.show()

# Conecta o botão à função de atualização
botao_atualizar.on_click(atualizar)

# Exibe os controles na tela
display(seletor_editora, seletor_ano, botao_atualizar, area_grafico)
````

# Exemplo 6 - Criação da Rede de Colaboração

O código a seguir simula uma base de referências com listas de autores, constrói uma rede de coautoria em que cada autor vira um nó e cada par de coautores vira uma aresta, calcula uma disposição espacial (spring layout) e exibe a rede em um gráfico interativo do Plotly com nós rotulados e conexões entre autores.

````python
import pandas as pd
import networkx as nx
import plotly.graph_objects as go

# 1. Simulação de dados extraídos das referências do artigo (Exemplo)
# Preencher o dicionário com os dados reais do PDF
data = {
    'Artigo_Citado': ['Anjum et al.', 'Albahri et al.', 'Ali et al.', 'Azimi et al.', 'Librantz et al.'],
    'Autores': [
        ['Anjum', 'Ahmed', 'Khan', 'Ahmad'],
        ['Albahri', 'Zaidan', 'Alsalem', 'Mohsin'],
        ['Ali', 'Khand', 'Kwak', 'Islam'],
        ['Azimi', 'Pahikkala', 'Rahmani', 'Liljeberg'],
        ['Librantz', 'Costa', 'Spinola', 'Oliveira Neto']
    ]
}

# 2. Construção da Rede
G = nx.Graph()
for autores in data['Autores']:
    for i in range(len(autores)):
        for j in range(i + 1, len(autores)):
            G.add_edge(autores[i], autores[j])

pos = nx.spring_layout(G)

# 3. Criação das Bordas (Edges)
edge_x = []
edge_y = []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

edge_trace = go.Scatter(x=edge_x, y=edge_y, line=dict(width=0.5, color='#888'), hoverinfo='none', mode='lines')

# 4. Criação dos Nós (Nodes)
node_x = []
node_y = []
for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)

node_trace = go.Scatter(
    x=node_x, y=node_y, mode='markers+text',
    text=[node for node in G.nodes()],
    textposition="bottom center",
    hoverinfo='text',
    marker=dict(showscale=True, colorscale='YlGnBu', size=15, color=[], line_width=2)
)

# 5. Dashboard com Plotly
fig = go.Figure(data=[edge_trace, node_trace],
             layout=go.Layout(
                title='Rede de Colaboração Acadêmica - Referências do Artigo',
                showlegend=False,
                hovermode='closest',
                margin=dict(b=20,l=5,r=5,t=40),
                xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                yaxis=dict(showgrid=False, zeroline=False, showticklabels=False))
                )

fig.show()
````

# Exemplo 7 - Metadados a partir da API da Crossref (Lista Fixa)

O código consulta a API da Crossref para obter os metadados de um artigo a partir do DOI informado e, em seguida, monta um grafo de relações entre autores para exibir uma rede interativa no Plotly.

Na prática, ele faz o seguinte: define uma função que faz uma requisição HTTP ao endpoint https://api.crossref.org/works/{doi} e retorna o JSON com os metadados quando a resposta é bem sucedida. Depois, embora o comentário indique que seria possível iterar sobre metadata['reference'] para extrair referências reais, o exemplo não faz essa extração e usa uma lista fixa de conexões entre nomes (uma simulação de vínculos de colaboração e influência).

Por fim, cria um grafo com NetworkX, calcula posições com spring_layout, converte arestas e nós em coordenadas, e plota a rede no Plotly, com linhas representando conexões e rótulos com os nomes dos autores.

````python
import requests
import pandas as pd
import networkx as nx
import plotly.graph_objects as go

def get_metadata(doi):
    url = f"https://api.crossref.org/works/{doi}"
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()['message']
    return None

# DOI do artigo fornecido no PDF
DOI_ARTIGO = "10.17013/risti.47.87-108"
metadata = get_metadata(DOI_ARTIGO)

# Extração de referências e autores (Simulação de processamento de rede)
# Na prática, os alunos iterariam sobre metadata['reference']
# Abaixo, estruturamos a lógica para o Dashboard
nodes = []
edges = []

# Exemplo de conexões entre autores principais e citados no artigo
conexoes = [
    ("Edson Melo de Souza", "Erika Midori Kinjo"),
    ("Edson Melo de Souza", "André Librantz"),
    ("Erika Midori Kinjo", "Fábio Cosme"),
    ("André Librantz", "J. Pearl"),
    ("Edson Melo de Souza", "Mittelstadt"),
    ("Fábio Cosme", "Anjum")
]

G = nx.Graph()
G.add_edges_from(conexoes)
pos = nx.spring_layout(G)

# --- Construção do Dashboard Interativo ---
edge_x, edge_y = [], []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

node_x, node_y, node_text = [], [], []
for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    node_text.append(node)

fig = go.Figure()

# Linhas da rede
fig.add_trace(go.Scatter(x=edge_x, y=edge_y, line=dict(width=1, color='#888'), hoverinfo='none', mode='lines'))

# Pontos (Autores)
fig.add_trace(go.Scatter(
    x=node_x, y=node_y, mode='markers+text',
    text=node_text, textposition="top center",
    marker=dict(size=20, color='skyblue', line_width=2)
))

fig.update_layout(
    title="Dashboard: Rede de Influência e Colaboração (via Crossref API)",
    showlegend=False,
    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
)

fig.show()
````

# Exemplo 8 - Metadados a partir da CrossRef (Dinamicamente)

O código consulta a API da Crossref usando um DOI e tenta extrair informações bibliográficas básicas e a lista de referências do artigo.

Ele define uma função que monta a URL do endpoint works/{doi}, faz a requisição HTTP com um cabeçalho User-Agent (boa prática para identificar a aplicação) e, se a resposta vier com status 200, acessa o campo message do JSON retornado. A partir daí, imprime o título do artigo (title[0]) e o nome da revista (container-title[0]). Em seguida, verifica se existe o campo reference nos metadados e, se existir, informa quantas referências foram encontradas e retorna essa lista.

No final, a função é executada para o DOI indicado e o conteúdo retornado (a lista de referências, ou None) é impresso.

````python
import requests
import json

def extrair_dados_crossref(doi):
    # Endpoint da API Crossref
    url = f"https://api.crossref.org/works/{doi}"

    # Cabeçalho para boas práticas (identificando o usuário)
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:seu-email@faculdade.edu.br)'}

    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        dados = response.json()['message']

        print(f"Título: {dados['title'][0]}")
        print(f"Revista: {dados['container-title'][0]}")

        # Extraindo a lista de referências citadas
        if 'reference' in dados:
            print(f"\nTotal de referências encontradas: {len(dados['reference'])}")
            return dados['reference']
    else:
        print("Erro ao acessar a API.")
        return None

# Executando com o DOI do artigo fornecido
referencias = extrair_dados_crossref("10.1007/s11192-022-04315-8")

print(json.dumps(referencias, indent=2, ensure_ascii=False))
````

# Exemplo 9 - Dashboard Bibliométrico: Rede de Coautoria e Citações via Crossref

O código consulta a API da Crossref para obter os metadados reais de um artigo a partir do DOI. Com esses metadados, ele constrói uma rede em NetworkX com dois tipos de relações: coautoria entre os autores do artigo principal (todos conectados entre si, com peso maior) e ligações de citação do primeiro autor do artigo para os autores identificados nas referências (peso menor, com uma extração simples do nome quando a referência é não estruturada). Em seguida, calcula o layout da rede com spring layout e gera um dashboard interativo no Plotly, exibindo nós e arestas, com autores principais destacados por tamanho maior e a visualização rotulada com o título do artigo.

````python
import requests
import pandas as pd
import networkx as nx
import plotly.graph_objects as go

def get_real_metadata(doi):
    url = f"https://api.crossref.org/works/{doi}"
    # É boa prática incluir um e-mail para a API da Crossref
    headers = {'User-Agent': 'TrabalhoAcademico/1.0 (mailto:prof.edson.melo@gmail.com)'}
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()['message']
    return None

# 1. Obtenção dos dados reais
DOI_ARTIGO = "10.1007/s11192-022-04315-8"
metadata = get_real_metadata(DOI_ARTIGO)

G = nx.Graph()

if metadata:
    # 2. Processar Autores do Artigo Principal
    # Conecta todos os co-autores entre si (Rede de Colaboração Direta)
    autores_principais = [f"{a.get('given', '')} {a.get('family', '')}".strip() for a in metadata.get('author', [])]
    for i in range(len(autores_principais)):
        for j in range(i + 1, len(autores_principais)):
            G.add_edge(autores_principais[i], autores_principais[j], weight=2)

    # 3. Processar Referências (Rede de Citação)
    # Aqui conectamos o primeiro autor do artigo principal aos autores citados
    autor_primeiro = autores_principais[0] if autores_principais else "Autor Principal"

    referencias = metadata.get('reference', [])
    for ref in referencias:
        # Tenta extrair o autor da citação
        autor_citado = None
        if 'author' in ref:
            autor_citado = ref['author']
        elif 'unstructured' in ref:
            # Extração simples: assume que o nome do autor está antes do ano ou primeira vírgula
            autor_citado = ref['unstructured'].split(',')[0].split('(')[0].strip()

        if autor_citado and len(autor_citado) < 50: # Filtro para evitar frases longas
            G.add_edge(autor_primeiro, autor_citado, weight=1)

# 4. Configuração do Layout da Rede
pos = nx.spring_layout(G, k=0.5, iterations=50)

# --- Construção do Dashboard Interativo com Plotly ---
edge_x, edge_y = [], []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

node_x, node_y, node_text, node_size = [], [], [], []
for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    node_text.append(node)
    # Diferencia o tamanho: autores principais são maiores
    size = 25 if node in autores_principais else 12
    node_size.append(size)

fig = go.Figure()

# Adiciona as conexões
fig.add_trace(go.Scatter(
    x=edge_x, y=edge_y,
    line=dict(width=0.5, color='#A5A5A5'),
    hoverinfo='none',
    mode='lines'
))

# Adiciona os autores
fig.add_trace(go.Scatter(
    x=node_x, y=node_y,
    mode='markers+text',
    text=node_text,
    textposition="top center",
    hoverinfo='text',
    marker=dict(
        showscale=True,
        colorscale='Viridis',
        size=node_size,
        color=node_size, # Cor baseada na importância/tamanho
        line_width=2)
))

fig.update_layout(
    title=f"Dashboard Real: Rede de Colaboração e Citações<br>Artigo: {metadata.get('title', [''])[0]}",
    titlefont_size=16,
    showlegend=False,
    hovermode='closest',
    margin=dict(b=20, l=5, r=5, t=60),
    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
)

fig.show()
````

# Exemplo 10 - Rede de Coautoria e Base de Citações via Crossref

O código consulta a API da Crossref para obter os metadados reais do artigo identificado pelo DOI informado. A partir desses metadados, ele constrói uma rede em NetworkX com dois grupos de nós: autores principais do artigo (marcados como “Principal”, em azul e maiores) e autores extraídos das referências do artigo (marcados como “Citado”, em vermelho e menores). Em seguida, ele liga todos os coautores entre si e também conecta cada autor principal a cada autor citado nas referências, formando uma rede de correlação entre autoria do artigo e sua base de citações. Por fim, calcula um layout do grafo e exibe um dashboard interativo no Plotly com nós rotulados e arestas representando as conexões.

````python
import requests
import pandas as pd
import networkx as nx
import plotly.graph_objects as go

def get_real_metadata(doi):
    url = f"https://api.crossref.org/works/{doi}"
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:prof.edson.melo@gmail.com)'}
    try:
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            return response.json()['message']
    except Exception as e:
        print(f"Erro na conexão: {e}")
    return None

# 1. Configuração e Busca Real (Artigo: Modelagem Bayesiana em IoT Saúde)
DOI_ARTIGO = "10.1016/j.joi.2017.03.003"
metadata = get_real_metadata(DOI_ARTIGO)

G = nx.Graph()

if metadata:
    # 2. Processar Autores Principais (Edson Melo, Erika Kinjo, etc.)
    autores_principais = [f"{a.get('given', '')} {a.get('family', '')}".strip() for a in metadata.get('author', [])]

    for autor in autores_principais:
        G.add_node(autor, categoria='Principal', cor='royalblue', tamanho=25)

    # Conecta os coautores (Correlação interna)
    for i in range(len(autores_principais)):
        for j in range(i + 1, len(autores_principais)):
            G.add_edge(autores_principais[i], autores_principais[j])

    # 3. Processar Referências (Rede de Citação Real)
    referencias = metadata.get('reference', [])
    for ref in referencias:
        autor_citado = None
        if 'author' in ref:
            autor_citado = ref['author']
        elif 'unstructured' in ref:
            # Extrai o sobrenome/nome antes da vírgula ou parênteses
            autor_citado = ref['unstructured'].split(',')[0].split('(')[0].strip()

        if autor_citado and 3 < len(autor_citado) < 40:
            if autor_citado not in G:
                G.add_node(autor_citado, categoria='Citado', cor='lightcoral', tamanho=12)

            # Conecta todos os autores principais à referência citada
            for principal in autores_principais:
                G.add_edge(principal, autor_citado)

# 4. Layout e Posição
pos = nx.spring_layout(G, k=0.5, iterations=100)

# 5. Preparação das Arestas (Linhas)
edge_x, edge_y = [], []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

# 6. Preparação dos Nós (Pontos e Nomes)
node_x, node_y, node_text, node_color, node_size = [], [], [], [], []
for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    node_text.append(node) # Este é o nome que aparecerá no grafo
    node_color.append(G.nodes[node]['cor'])
    node_size.append(G.nodes[node]['tamanho'])

# 7. Construção do Dashboard
fig = go.Figure()

# Adiciona as linhas de conexão
fig.add_trace(go.Scatter(
    x=edge_x, y=edge_y,
    line=dict(width=0.5, color='#888'),
    hoverinfo='none', mode='lines'
))

# Adiciona os nós com nomes visíveis
fig.add_trace(go.Scatter(
    x=node_x, y=node_y,
    mode='markers+text', # 'text' ativa a exibição dos nomes
    text=node_text,
    textposition="top center", # Garante que o nome fique acima do ponto
    hoverinfo='text',
    marker=dict(
        size=node_size,
        color=node_color,
        line=dict(width=2, color='white')
    ),
    textfont=dict(size=10)
))

fig.update_layout(
    title=f"Rede de Correlação: {metadata.get('title', [''])[0]}",
    showlegend=False,
    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    plot_bgcolor='white',
    margin=dict(t=80)
)

fig.show()
````

# Exemplo 11 - Análise da Atualidade e Autores Dominantes nas Referências

O código consulta a API da Crossref para obter as referências (bibliografia) de um artigo via DOI, extrai de cada referência o ano, um autor identificável e um indicador de citações, monta um DataFrame com essa base teórica classificada em “clássica” (antes de 2015) e “contemporânea”, gera dois gráficos (histograma temporal da bibliografia e ranking dos 5 autores mais frequentes nas referências) e, ao final, imprime métricas resumo como média de citações, referência mais citada e total de referências processadas.

````python
import requests
import pandas as pd
import plotly.express as px
import re
from collections import Counter

def get_metadata(doi):
    url = f"https://api.crossref.org/works/{doi}"
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:prof.edson.melo@gmail.com)'}
    try:
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            return response.json()['message']
    except:
        return None
    return None

# 1. Configuração e Busca Real (DOI do artigo RISTI)
DOI_ARTIGO = "10.1007/s11192-022-04315-8"
metadata = get_metadata(DOI_ARTIGO)
referencias = metadata.get('reference', []) if metadata else []

lista_dados = []
lista_autores = []

# 2. Processamento das Referências
for ref in referencias:
    # Extração robusta do Ano
    ano = None
    if 'year' in ref:
        ano = str(ref['year'])
    elif 'unstructured' in ref:
        match = re.search(r'\b(19|20)\d{2}\b', ref['unstructured'])
        if match:
            ano = match.group()

    # Extração de Autores
    autor = "Não Identificado"
    if 'author' in ref:
        autor = ref['author']
    elif 'unstructured' in ref:
        autor = ref['unstructured'].split(',')[0].split('(')[0].strip()

    # Impacto (Citações que a referência recebeu)
    # A Crossref fornece isso no campo 'is-referenced-by-count' dentro do objeto da referência
    citacoes = ref.get('is-referenced-by-count', 0)

    if ano and len(ano) == 4:
        ano_int = int(ano)
        lista_autores.append(autor)
        lista_dados.append({
            'Ano': ano_int,
            'Autor': autor,
            'Citações': citacoes,
            'Classificação': 'Clássica (Pré-2015)' if ano_int < 2015 else 'Contemporânea'
        })

df = pd.DataFrame(lista_dados)

# --- VISUALIZAÇÃO 1: Histograma de Atualidade ---
df_contagem = df.groupby(['Ano', 'Classificação']).size().reset_index(name='Frequência')

fig_hist = px.bar(
    df_contagem,
    x='Ano', y='Frequência', color='Classificação',
    title="Histograma de Atualidade: Evolução da Base Teórica",
    color_discrete_map={'Clássica (Pré-2015)': '#EF553B', 'Contemporânea': '#636EFA'}
)
fig_hist.update_layout(bargap=0.2)
fig_hist.show()

# --- VISUALIZAÇÃO 2: Top 5 Autores mais Citados ---
top_5_autores = pd.DataFrame(Counter(lista_autores).most_common(5), columns=['Autor', 'Frequência'])
fig_autores = px.bar(
    top_5_autores, x='Frequência', y='Autor', orientation='h',
    title="Top 5 Autores mais Frequentes nas Referências",
    color='Frequência', color_continuous_scale='Viridis'
)
fig_autores.update_yaxes(autorange="reversed") # Deixa o mais citado no topo
fig_autores.show()

# --- INDICADORES FINAIS ---
print(f"\n" + "="*30)
print(f"RESUMO BIBLIOMÉTRICO REAL")
print(f"="*30)
print(f"Média de Citações dos artigos da base: {df['Citações'].mean():.2f}")
print(f"Artigo mais citado da bibliografia: {df.loc[df['Citações'].idxmax(), 'Autor']} ({df['Citações'].max()} citações)")
print(f"Total de referências processadas: {len(df)}")
````

# Exemplo 12 - Análise da Atualidade e Impacto da Base Teórica via Crossref

O código consulta a API da Crossref para obter as referências de um artigo (via DOI). Para cada referência, se houver DOI, ele busca metadados completos para recuperar o número de citações e o ano de publicação; quando isso falha ou não há DOI, ele faz extração heurística do ano e do autor a partir do texto “unstructured”. Com os dados consolidados, ele classifica as referências em clássicas (pré-2015) e contemporâneas, gera um histograma temporal da base teórica e um ranking dos autores mais recorrentes, e imprime indicadores finais como média de citações, referência mais citada e total de referências analisadas.

````python
import requests
import pandas as pd
import plotly.express as px
import re
from collections import Counter
import time

def get_metadata(doi):
    """Busca metadados completos de um DOI específico"""
    url = f"https://api.crossref.org/works/{doi}"
    # Identificação recomendada para não ser bloqueado pela API
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:prof.edson.melo@gmail.com)'}
    try:
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            return response.json()['message']
    except:
        return None
    return None

# 1. Busca inicial do artigo principal (RISTI)
DOI_ARTIGO = "10.1007/s11192-022-04315-8"
print(f"Buscando referências do artigo principal...")
metadata_principal = get_metadata(DOI_ARTIGO)
referencias_brutas = metadata_principal.get('reference', []) if metadata_principal else []

lista_final = []
lista_autores = []

print(f"Encontradas {len(referencias_brutas)} referências. Iniciando coleta de citações individuais...")

# 2. Loop para processar cada referência e buscar seu impacto real
for i, ref in enumerate(referencias_brutas):
    doi_ref = ref.get('DOI')
    ano = None
    autor = "Não Identificado"
    qtd_citacoes = 0

    # Se a referência tiver DOI, buscamos os dados atualizados de impacto
    if doi_ref:
        detalhes = get_metadata(doi_ref)
        if detalhes:
            qtd_citacoes = detalhes.get('is-referenced-by-count', 0)
            # Tenta pegar o ano da data de publicação oficial
            try:
                ano = detalhes.get('published-print', detalhes.get('issued')).get('date-parts')[0][0]
            except:
                pass

            # Tenta pegar o autor principal estruturado
            if detalhes.get('author'):
                autor = detalhes['author'][0].get('family', 'Autor Desconhecido')

    # Fallback: Se não tiver DOI ou os dados acima falharem, tentamos extrair do texto bruto
    if not ano or autor == "Não Identificado":
        if 'year' in ref:
            ano = str(ref['year'])
        elif 'unstructured' in ref:
            # Busca o ano com Regex (ex: 2019)
            match_ano = re.search(r'\b(19|20)\d{2}\b', ref['unstructured'])
            if match_ano: ano = match_ano.group()

            # Pega o primeiro nome antes da vírgula no texto bruto
            if autor == "Não Identificado":
                autor = ref['unstructured'].split(',')[0].split('(')[0].strip()

    # Só adicionamos ao relatório se tivermos o ano para o histograma
    if ano:
        ano_int = int(ano)
        lista_autores.append(autor)
        lista_final.append({
            'Ano': ano_int,
            'Autor': autor,
            'Citações': qtd_citacoes,
            'Classificação': 'Clássica (Pré-2015)' if ano_int < 2015 else 'Contemporânea'
        })

    # Feedback para o usuário não achar que o código travou
    if (i + 1) % 10 == 0:
        print(f"Processados {i+1} de {len(referencias_brutas)}...")

# 3. Geração dos Gráficos
df = pd.DataFrame(lista_final)

# Histograma de Atualidade
df_contagem = df.groupby(['Ano', 'Classificação']).size().reset_index(name='Frequência')
fig_hist = px.bar(
    df_contagem, x='Ano', y='Frequência', color='Classificação',
    title="Histograma de Atualidade: Evolução da Base Teórica",
    color_discrete_map={'Clássica (Pré-2015)': '#EF553B', 'Contemporânea': '#636EFA'}
)
fig_hist.show()

# Top Autores
top_autores = pd.DataFrame(Counter(lista_autores).most_common(5), columns=['Autor', 'Frequência'])
fig_top = px.bar(top_autores, x='Frequência', y='Autor', orientation='h', title="Autores mais recorrentes na base")
fig_top.update_yaxes(autorange="reversed")
fig_top.show()

# 4. Resultados Numéricos
print(f"\n" + "="*40)
print(f"RESULTADOS DA ANÁLISE REAL DE IMPACTO")
print(f"="*40)
print(f"Média de citações (Impacto da Base): {df['Citações'].mean():.2f}")
print(f"Referência mais influente: {df.loc[df['Citações'].idxmax(), 'Autor']} ({df['Citações'].max()} citações)")
print(f"Total de referências analisadas: {len(df)}")
````

````python
import requests
import pandas as pd
import plotly.express as px
import re
from collections import Counter
import time

def get_metadata(doi):
    """Busca metadados completos de um DOI específico via Crossref"""
    url = f"https://api.crossref.org/works/{doi}"
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:prof.edson.melo@gmail.com)'}
    try:
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            return response.json()['message']
    except:
        return None
    return None

# 1. Configuração e Busca do Artigo Principal (RISTI)
DOI_ARTIGO = "10.3390/systems13010037"
print(f"Buscando referências do artigo: {DOI_ARTIGO}")
metadata_principal = get_metadata(DOI_ARTIGO)
referencias_brutas = metadata_principal.get('reference', []) if metadata_principal else []

lista_final = []
print(f"Total de {len(referencias_brutas)} referências encontradas. Iniciando coleta profunda de impacto...")

# 2. Processamento das Referências e Coleta de Citações Reais
for i, ref in enumerate(referencias_brutas):
    doi_ref = ref.get('DOI')
    ano = None
    autor = "Não Identificado"
    qtd_citacoes = 0

    # Se houver DOI, buscamos o impacto real e dados estruturados
    if doi_ref:
        detalhes = get_metadata(doi_ref)
        if detalhes:
            qtd_citacoes = detalhes.get('is-referenced-by-count', 0)
            try:
                # Tenta capturar o ano de publicação oficial
                data = detalhes.get('published-print', detalhes.get('issued'))
                ano = data.get('date-parts')[0][0]
            except:
                pass

            # Captura o sobrenome do autor principal
            if detalhes.get('author'):
                autor = detalhes['author'][0].get('family', 'Autor Desconhecido')

    # Fallback: Se não houver DOI ou dados estruturados, tenta extrair do texto bruto
    if not ano or autor == "Não Identificado":
        if 'year' in ref:
            ano = str(ref['year'])
        elif 'unstructured' in ref:
            match_ano = re.search(r'\b(19|20)\d{2}\b', ref['unstructured'])
            if match_ano: ano = match_ano.group()

            if autor == "Não Identificado":
                # Tenta isolar o nome antes da primeira vírgula ou parênteses
                autor = ref['unstructured'].split(',')[0].split('(')[0].strip()

    if ano:
        lista_final.append({
            'Ano': int(ano),
            'Autor': autor,
            'Citações': qtd_citacoes,
            'Classificação': 'Clássica (Pré-2015)' if int(ano) < 2015 else 'Contemporânea'
        })

    if (i + 1) % 10 == 0:
        print(f"Processados {i+1} de {len(referencias_brutas)}...")

# 3. Preparação dos Dados
df = pd.DataFrame(lista_final)

# --- VISUALIZAÇÃO 1: Histograma de Atualidade ---
df_contagem = df.groupby(['Ano', 'Classificação']).size().reset_index(name='Frequência')
fig_hist = px.bar(
    df_contagem, x='Ano', y='Frequência', color='Classificação',
    title="Evolução Temporal da Base Teórica (Clássica vs Contemporânea)",
    color_discrete_map={'Clássica (Pré-2015)': '#EF553B', 'Contemporânea': '#636EFA'}
)
fig_hist.show()

# --- VISUALIZAÇÃO 2: Frequência de Autores na Bibliografia ---
top_frequencia = pd.DataFrame(Counter(df['Autor']).most_common(10), columns=['Autor', 'Aparições'])
fig_freq = px.bar(top_frequencia, x='Aparições', y='Autor', orientation='h',
                  title="Top 10 Autores mais Frequentes na Bibliografia")
fig_freq.update_yaxes(autorange="reversed")
fig_freq.show()

# --- 4. RELATÓRIO FINAL: OS 10 MAIS CITADOS (IMPACTO REAL) ---
# Aqui somamos as citações que cada autor possui na base da Crossref
df_impacto = df.groupby('Autor')['Citações'].sum().sort_values(ascending=False).head(10).reset_index()

print("\n" + "="*50)
print(f"ANÁLISE DE IMPACTO: TOP 10 AUTORES MAIS CITADOS")
print(f"(Baseado no impacto global das obras citadas)")
print("="*50)

for index, row in df_impacto.iterrows():
    print(f"{index+1}º - {row['Autor']}: {int(row['Citações'])} citações totais")

print("="*50)

# 4. Resultados Numéricos
print(f"\n" + "="*40)
print(f"RESULTADOS DA ANÁLISE REAL DE IMPACTO")
print(f"="*40)
print(f"Média de citações (Impacto da Base): {df['Citações'].mean():.2f}")
print(f"Referência mais influente: {df.loc[df['Citações'].idxmax(), 'Autor']} ({df['Citações'].max()} citações)")
print(f"Total de referências analisadas: {len(df)}")
````

# Exemplo 14 - Ranking de Veículos Citados nas Referências via Crossref

O código consulta a API da Crossref para obter as referências de um artigo pelo DOI e, para cada referência, tenta identificar o veículo de publicação (periódico ou conferência). Quando a referência tem DOI, ele busca os metadados completos e extrai o nome do veículo pelo campo container-title. Se não houver DOI ou a API não retornar esse campo, ele tenta inferir o veículo a partir do texto “unstructured” usando regras simples para alguns nomes conhecidos. Com os veículos coletados, ele monta um ranking dos 10 mais frequentes, exibe um gráfico de barras horizontal com Plotly e imprime o ranking final no console.

````python
import requests
import pandas as pd
import plotly.express as px
import re
from collections import Counter

def get_metadata(doi):
    """Busca metadados completos via Crossref"""
    url = f"https://api.crossref.org/works/{doi}"
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:prof.edson.melo@gmail.com)'}
    try:
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            return response.json()['message']
    except:
        return None
    return None

# 1. Busca do Artigo Principal
DOI_ARTIGO = "10.1007/s11192-022-04315-8"
print(f"Iniciando análise de fontes para o DOI: {DOI_ARTIGO}")
metadata_principal = get_metadata(DOI_ARTIGO)
referencias_brutas = metadata_principal.get('reference', []) if metadata_principal else []

lista_fontes = []

print(f"Analisando {len(referencias_brutas)} referências...")

# 2. Coleta de Nomes de Periódicos/Conferências
for i, ref in enumerate(referencias_brutas):
    fonte = "Não Identificada"
    doi_ref = ref.get('DOI')

    # Tentativa 1: Dados Estruturados via API (Mais preciso)
    if doi_ref:
        detalhes = get_metadata(doi_ref)
        if detalhes and detalhes.get('container-title'):
            fonte = detalhes['container-title'][0]

    # Tentativa 2: Extração do texto bruto (unstructured) caso falhe a API
    if fonte == "Não Identificada" and 'unstructured' in ref:
        # Geralmente o nome da revista vem em itálico ou após o título;
        # Aqui usamos uma lógica simples de busca por nomes conhecidos ou padrões
        texto = ref['unstructured']
        # Tenta identificar nomes comuns de journals (IEEE, Sensors, Nature, etc)
        if "IEEE" in texto: fonte = "IEEE (Diversos)"
        elif "Sensors" in texto: fonte = "Sensors Journal"
        elif "RISTI" in texto: fonte = "RISTI"
        elif "Future Generation Computer Systems" in texto: fonte = "Future Generation Computer Systems"
        elif "Ethics and Information Technology" in texto: fonte = "Ethics and Information Technology"

    if fonte != "Não Identificada":
        lista_fontes.append(fonte)

    if (i + 1) % 10 == 0:
        print(f"Processados {i+1} de {len(referencias_brutas)}...")

# 3. Processamento do Ranking
ranking_fontes = pd.DataFrame(Counter(lista_fontes).most_common(10), columns=['Veículo', 'Frequência'])

# 4. Visualização: Onde os autores mais publicam
fig_fontes = px.bar(
    ranking_fontes,
    x='Frequência',
    y='Veículo',
    orientation='h',
    title="Top 10 Veículos de Publicação mais Citados",
    labels={'Frequência': 'Número de Artigos Citados', 'Veículo': 'Nome do Periódico/Conferência'},
    color='Frequência',
    color_continuous_scale='Plasma'
)
fig_fontes.update_yaxes(autorange="reversed")
fig_fontes.show()

# 5. Relatório Final de Fontes
print("\n" + "="*50)
print("RANKING DE VEÍCULOS (JOURNALS/CONFERENCES)")
print("="*50)
for index, row in ranking_fontes.iterrows():
    print(f"{index+1}º - {row['Veículo']} ({row['Frequência']} artigos)")
print("="*50)
````

# Exemplo 15 - Dashboard Bibliométrico Completo: Rede de Coautoria, Atualidade e Fontes da Base Teórica via Crossref

O código consulta a API da Crossref para obter os metadados de um artigo pelo DOI e, a partir das referências desse artigo, tenta enriquecer cada referência com DOI, ano, autor, fonte (periódico/veículo) e número de citações. Com esses dados, ele monta um DataFrame e também constrói uma rede em que os autores do artigo principal se conectam aos autores citados na bibliografia.

No final, ele gera três visualizações: uma rede interativa de coautoria e base teórica (Plotly + NetworkX), um histograma com a distribuição temporal das referências (atualidade da base teórica) e um gráfico de barras com os 10 veículos mais frequentes nas referências. Por fim, imprime um ranking com os 10 autores citados de maior impacto, somando as citações obtidas via Crossref.

````python
import requests
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import networkx as nx
import re
from collections import Counter

# --- CONFIGURAÇÕES E API ---
def get_metadata(doi):
    url = f"https://api.crossref.org/works/{doi}"
    headers = {'User-Agent': 'AtividadeAcademica/1.0 (mailto:prof.edson.melo@gmail.com)'}
    try:
        response = requests.get(url, headers=headers)
        return response.json()['message'] if response.status_code == 200 else None
    except:
        return None

# DOI do seu artigo fornecido
DOI_ARTIGO = "10.1007/s11192-022-04315-8"
print(f"Iniciando processamento completo para o artigo: {DOI_ARTIGO}...")

metadata_principal = get_metadata(DOI_ARTIGO)
referencias_brutas = metadata_principal.get('reference', []) if metadata_principal else []

lista_dados = []
G = nx.Graph()

# Extração de autores principais para a rede
autores_principais = [f"{a.get('given', '')} {a.get('family', '')}".strip() for a in metadata_principal.get('author', [])]
for autor in autores_principais:
    G.add_node(autor, categoria='Principal', cor='royalblue', tamanho=30)

# --- PROCESSAMENTO REAL ---
print(f"Processando {len(referencias_brutas)} referências. Por favor, aguarde...")

for i, ref in enumerate(referencias_brutas):
    doi_ref = ref.get('DOI')
    ano, autor, fonte, citacoes = None, "Não Identificado", "Não Identificada", 0

    if doi_ref:
        detalhes = get_metadata(doi_ref)
        if detalhes:
            citacoes = detalhes.get('is-referenced-by-count', 0)
            try:
                data = detalhes.get('published-print', detalhes.get('issued'))
                ano = data.get('date-parts')[0][0]
            except: pass
            if detalhes.get('author'): autor = detalhes['author'][0].get('family')
            if detalhes.get('container-title'): fonte = detalhes['container-title'][0]

    if not ano or autor == "Não Identificado":
        if 'year' in ref: ano = str(ref['year'])
        elif 'unstructured' in ref:
            m = re.search(r'\b(19|20)\d{2}\b', ref['unstructured'])
            if m: ano = m.group()
            autor = ref['unstructured'].split(',')[0].split('(')[0].strip()
            if "IEEE" in ref['unstructured']: fonte = "IEEE"
            elif "Sensors" in ref['unstructured']: fonte = "Sensors Journal"

    if ano:
        lista_dados.append({'Ano': int(ano), 'Autor': autor, 'Fonte': fonte, 'Citações': citacoes})
        # Adiciona na Rede
        if autor not in G: G.add_node(autor, categoria='Citado', cor='lightcoral', tamanho=12)
        for p in autores_principais: G.add_edge(p, autor)

    if (i + 1) % 15 == 0: print(f"Progresso: {i+1} referências analisadas...")

df = pd.DataFrame(lista_dados)

# --- 1. VISUALIZAÇÃO: REDE DE CORRELAÇÃO ---
pos = nx.spring_layout(G, k=0.4)
edge_x, edge_y = [], []
for edge in G.edges():
    x0, y0, x1, y1 = pos[edge[0]][0], pos[edge[0]][1], pos[edge[1]][0], pos[edge[1]][1]
    edge_x.extend([x0, x1, None]); edge_y.extend([y0, y1, None])

fig_rede = go.Figure(data=[
    go.Scatter(x=edge_x, y=edge_y, line=dict(width=0.5, color='#888'), hoverinfo='none', mode='lines'),
    go.Scatter(x=[pos[n][0] for n in G.nodes()], y=[pos[n][1] for n in G.nodes()],
               mode='markers+text', text=list(G.nodes()), textposition="top center",
               marker=dict(size=[G.nodes[n]['tamanho'] for n in G.nodes()],
                           color=[G.nodes[n]['cor'] for n in G.nodes()], line_width=2))
])
fig_rede.update_layout(title="Rede de Coautoria e Base Teórica", showlegend=False)
fig_rede.show()

# --- 2. VISUALIZAÇÃO: HISTOGRAMA DE ATUALIDADE ---
fig_hist = px.histogram(df, x="Ano", title="Evolução da Base Teórica", nbins=20, color_discrete_sequence=['#636EFA'])
fig_hist.show()

# --- 3. VISUALIZAÇÃO: ONDE OS AUTORES PUBLICAM ---
top_fontes = df['Fonte'].value_counts().head(10).reset_index()
fig_fontes = px.bar(top_fontes, x='count', y='Fonte', orientation='h', title="Top 10 Veículos de Publicação")
fig_fontes.show()

# --- 4. RELATÓRIO FINAL ---
print("\n" + "="*50 + "\nRANKING DE IMPACTO (TOP 10 AUTORES)\n" + "="*50)
print(df.groupby('Autor')['Citações'].sum().sort_values(ascending=False).head(10))
````

# Exemplo 16 - Dashboard bibliométrico integrado com rede de coautoria, temporalidade de referências, veículos mais citados e ranking de impacto via Crossref

O código lê um CSV no padrão típico de exportação do Scopus e faz uma padronização mínima de colunas como Year e DOI. Em seguida, constrói um grafo de coautoria usando NetworkX, onde cada nó é um autor e cada aresta representa uma coautoria, com peso igual ao número de vezes que a dupla aparece no conjunto. Esse grafo é projetado em 2D com spring layout e renderizado de forma interativa no Plotly, permitindo inspeção por hover.

Em paralelo, o código gera duas visões descritivas da base: um histograma com a distribuição dos anos (para avaliar a atualidade do corpus ou das referências, dependendo do que o seu CSV representa) e um gráfico de barras horizontais com os 10 veículos mais frequentes, obtidos por contagem do campo Source title.

Por fim, ele calcula um ranking de impacto por autor. Para cada DOI disponível no dataset, consulta a API do Crossref e captura o ```is-referenced-by-count```, que é usado como proxy de citações. Esse valor é somado ao “score” de todos os autores associados ao registro; ao final, o script ordena e seleciona os 10 autores com maior soma. Para reduzir tempo e chamadas repetidas, ele mantém um cache local em JSON com os DOIs já consultados. As quatro visualizações são combinadas em uma única figura Plotly com subplots, formando um dashboard único exibido com fig.show().

````python
import os
import json
import time
import math
import requests
import pandas as pd
import networkx as nx

import plotly.graph_objects as go
from plotly.subplots import make_subplots


# =========================
# Configurações principais
# =========================

CSV_PATH = "/content/dataset_exemplo.csv"  # ajuste se necessário
CROSSREF_MAILTO = "seu_email@dominio.com"   # recomendado pelo Crossref (User-Agent)
CROSSREF_CACHE_PATH = "crossref_cache.json"
CROSSREF_RATE_LIMIT_S = 0.25               # ~4 req/s. Ajuste se quiser ser mais conservador.
MAX_DOI_REQUESTS = None                    # ex.: 200 para testes. None = todos.

# Se você quiser usar o "Cited by" do Scopus como fallback quando DOI não existir:
USE_SCOPUS_CITED_BY_FALLBACK = True


# =========================
# Funções utilitárias
# =========================

def load_scopus_csv(path: str) -> pd.DataFrame:
    df = pd.read_csv(path)
    # Normalizações mínimas
    if "Year" in df.columns:
        df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")
    if "DOI" in df.columns:
        df["DOI"] = df["DOI"].astype(str).str.strip()
        df.loc[df["DOI"].isin(["nan", "None", ""]), "DOI"] = pd.NA
    return df


def parse_authors(authors_cell: str) -> list[str]:
    """
    Converte o padrão do Scopus "Sobrenome, I.; Sobrenome, I.I.; ..." em lista limpa.
    """
    if authors_cell is None or (isinstance(authors_cell, float) and math.isnan(authors_cell)):
        return []
    s = str(authors_cell).strip()
    if not s or s.lower() == "nan":
        return []
    parts = [p.strip() for p in s.split(";")]
    # remove vazios
    return [p for p in parts if p]


def safe_title(x) -> str:
    if x is None or (isinstance(x, float) and math.isnan(x)):
        return ""
    s = str(x).strip()
    return "" if s.lower() == "nan" else s


def load_crossref_cache(cache_path: str) -> dict:
    if os.path.exists(cache_path):
        try:
            with open(cache_path, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception:
            return {}
    return {}


def save_crossref_cache(cache_path: str, cache: dict) -> None:
    tmp = cache_path + ".tmp"
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(cache, f, ensure_ascii=False, indent=2)
    os.replace(tmp, cache_path)


def fetch_crossref_citations_for_doi(doi: str, session: requests.Session, mailto: str) -> int | None:
    """
    Retorna is-referenced-by-count (número de citações segundo Crossref), ou None se falhar.
    """
    url = f"https://api.crossref.org/works/{doi}"
    headers = {
        "User-Agent": f"BibliometriaDashboard/1.0 (mailto:{mailto})"
    }
    try:
        r = session.get(url, headers=headers, timeout=20)
        if r.status_code != 200:
            return None
        data = r.json()
        msg = data.get("message", {})
        cnt = msg.get("is-referenced-by-count", None)
        if isinstance(cnt, int):
            return cnt
        return None
    except Exception:
        return None


# =========================
# Métricas para os painéis
# =========================

def compute_top_sources(df: pd.DataFrame, top_n: int = 10) -> pd.DataFrame:
    if "Source title" not in df.columns:
        return pd.DataFrame(columns=["Source title", "count"])
    vc = df["Source title"].astype(str).replace("nan", pd.NA).dropna().str.strip().value_counts()
    out = vc.head(top_n).reset_index()
    out.columns = ["Source title", "count"]
    return out


def compute_year_distribution(df: pd.DataFrame) -> pd.Series:
    if "Year" not in df.columns:
        return pd.Series([], dtype="int64")
    return df["Year"].dropna().astype(int)


def build_coauthorship_graph(df: pd.DataFrame, min_edge_weight: int = 1) -> nx.Graph:
    """
    Grafo não direcionado: autores como nós, arestas para coautorias no mesmo registro.
    Peso = número de ocorrências de coautoria no dataset.
    """
    G = nx.Graph()
    if "Authors" not in df.columns:
        return G

    for _, row in df.iterrows():
        authors = parse_authors(row.get("Authors"))
        # adiciona nós
        for a in authors:
            if not G.has_node(a):
                G.add_node(a)

        # adiciona arestas (pares)
        for i in range(len(authors)):
            for j in range(i + 1, len(authors)):
                u, v = authors[i], authors[j]
                if G.has_edge(u, v):
                    G[u][v]["weight"] += 1
                else:
                    G.add_edge(u, v, weight=1)

    # filtra arestas fracas se necessário
    if min_edge_weight > 1:
        to_remove = [(u, v) for u, v, d in G.edges(data=True) if d.get("weight", 1) < min_edge_weight]
        G.remove_edges_from(to_remove)

    # remove isolados (opcional)
    isolates = list(nx.isolates(G))
    G.remove_nodes_from(isolates)
    return G


def compute_author_impact_ranking(
    df: pd.DataFrame,
    cache_path: str,
    mailto: str,
    rate_limit_s: float = 0.25,
    max_doi_requests: int | None = None,
    use_scopus_fallback: bool = True
) -> pd.DataFrame:
    """
    Para cada registro com DOI:
      - busca citações no Crossref (cacheado)
      - distribui esse "impacto" para todos os autores do registro (soma)
    Retorna top 10 autores por impacto acumulado.
    """
    cache = load_crossref_cache(cache_path)
    session = requests.Session()

    author_score = {}

    doi_list = []
    if "DOI" in df.columns:
        doi_list = df["DOI"].dropna().astype(str).str.strip().tolist()

    # remove duplicados mantendo ordem
    seen = set()
    unique_dois = []
    for d in doi_list:
        if d not in seen:
            unique_dois.append(d)
            seen.add(d)

    if max_doi_requests is not None:
        unique_dois = unique_dois[:max_doi_requests]

    # pré-carrega mapa DOI -> citações (Crossref/cache)
    doi_to_cites = {}
    for doi in unique_dois:
        if doi in cache and isinstance(cache[doi], int):
            doi_to_cites[doi] = cache[doi]
            continue

        cites = fetch_crossref_citations_for_doi(doi, session=session, mailto=mailto)
        if cites is not None:
            doi_to_cites[doi] = cites
            cache[doi] = cites
        else:
            cache[doi] = None
            doi_to_cites[doi] = None

        time.sleep(rate_limit_s)

    save_crossref_cache(cache_path, cache)

    # atribui impacto por autor
    for _, row in df.iterrows():
        authors = parse_authors(row.get("Authors"))
        if not authors:
            continue

        doi = row.get("DOI")
        cites = None

        if isinstance(doi, str) and doi.strip():
            cites = doi_to_cites.get(doi.strip(), None)

        if cites is None and use_scopus_fallback and "Cited by" in df.columns:
            # fallback: usa "Cited by" do Scopus se existir e for numérico
            sc = row.get("Cited by")
            try:
                if pd.notna(sc):
                    sc_int = int(sc)
                    cites = sc_int
            except Exception:
                pass

        if cites is None:
            continue

        for a in authors:
            author_score[a] = author_score.get(a, 0) + int(cites)

    ranking = (
        pd.DataFrame({"Author": list(author_score.keys()), "Impact": list(author_score.values())})
        .sort_values("Impact", ascending=False)
        .head(10)
        .reset_index(drop=True)
    )
    return ranking


# =========================
# Construção do dashboard
# =========================

def make_network_trace(G: nx.Graph) -> tuple[list[go.Scatter], list[go.Scatter]]:
    """
    Retorna (edge_traces, node_trace) para Plotly.
    """
    if G.number_of_nodes() == 0:
        # vazio
        node_trace = go.Scatter(
            x=[], y=[],
            mode="markers+text",
            text=[],
            hovertext=[],
            hoverinfo="text",
            textposition="top center",
            marker=dict(size=10),
            showlegend=False
        )
        return [], [node_trace]

    # layout
    pos = nx.spring_layout(G, seed=42, k=None)

    # arestas: uma trace única com segmentos
    edge_x, edge_y = [], []
    edge_hover = []
    for u, v, d in G.edges(data=True):
        x0, y0 = pos[u]
        x1, y1 = pos[v]
        edge_x += [x0, x1, None]
        edge_y += [y0, y1, None]
        w = d.get("weight", 1)
        edge_hover.append(f"{u} ↔ {v} | weight={w}")

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        mode="lines",
        hoverinfo="none",
        line=dict(width=1),
        showlegend=False
    )

    # nós
    nodes = list(G.nodes())
    deg = dict(G.degree())
    node_x = [pos[n][0] for n in nodes]
    node_y = [pos[n][1] for n in nodes]

    node_size = []
    node_text = []
    node_hover = []
    for n in nodes:
        d = deg.get(n, 0)
        node_size.append(6 + 2 * math.sqrt(d))
        node_text.append("")  # sem rótulo fixo para não poluir
        node_hover.append(f"{n}<br>degree={d}")

    node_trace = go.Scatter(
        x=node_x, y=node_y,
        mode="markers",
        hovertext=node_hover,
        hoverinfo="text",
        marker=dict(size=node_size, opacity=0.9),
        showlegend=False
    )

    return [edge_trace], [node_trace]


def build_dashboard_figure(df: pd.DataFrame) -> go.Figure:
    # Dados
    years = compute_year_distribution(df)
    top_sources = compute_top_sources(df, top_n=10)

    G = build_coauthorship_graph(df, min_edge_weight=1)
    edge_traces, node_traces = make_network_trace(G)

    ranking = compute_author_impact_ranking(
        df=df,
        cache_path=CROSSREF_CACHE_PATH,
        mailto=CROSSREF_MAILTO,
        rate_limit_s=CROSSREF_RATE_LIMIT_S,
        max_doi_requests=MAX_DOI_REQUESTS,
        use_scopus_fallback=USE_SCOPUS_CITED_BY_FALLBACK
    )

    # Layout 2x2
    fig = make_subplots(
        rows=2, cols=2,
        specs=[
            [{"type": "xy"}, {"type": "xy"}],
            [{"type": "xy"}, {"type": "table"}],
        ],
        subplot_titles=(
            "Rede de coautoria (interativa)",
            "Distribuição temporal (ano das referências)",
            "Top 10 veículos (Source title)",
            "Top 10 autores por impacto (Crossref)"
        ),
        horizontal_spacing=0.10,
        vertical_spacing=0.15
    )

    # (1) Rede: edges + nodes no (1,1)
    for tr in edge_traces:
        fig.add_trace(tr, row=1, col=1)
    for tr in node_traces:
        fig.add_trace(tr, row=1, col=1)

    fig.update_xaxes(visible=False, row=1, col=1)
    fig.update_yaxes(visible=False, row=1, col=1)

    # (2) Histograma de anos
    if len(years) > 0:
        fig.add_trace(
            go.Histogram(
                x=years,
                nbinsx=min(30, max(5, years.nunique())),
                hovertemplate="Ano=%{x}<br>Contagem=%{y}<extra></extra>",
                showlegend=False
            ),
            row=1, col=2
        )
        fig.update_xaxes(title_text="Ano", row=1, col=2)
        fig.update_yaxes(title_text="Contagem", row=1, col=2)
    else:
        fig.add_trace(go.Scatter(x=[], y=[], showlegend=False), row=1, col=2)

    # (3) Barras top fontes
    if not top_sources.empty:
        fig.add_trace(
            go.Bar(
                x=top_sources["count"],
                y=top_sources["Source title"],
                orientation="h",
                hovertemplate="%{y}<br>Contagem=%{x}<extra></extra>",
                showlegend=False
            ),
            row=2, col=1
        )
        fig.update_xaxes(title_text="Contagem", row=2, col=1)
        fig.update_yaxes(title_text="", row=2, col=1)
    else:
        fig.add_trace(go.Scatter(x=[], y=[], showlegend=False), row=2, col=1)

    # (4) Tabela ranking
    if ranking.empty:
        ranking = pd.DataFrame({"Author": [], "Impact": []})

    fig.add_trace(
        go.Table(
            header=dict(
                values=["Autor", "Impacto (soma de citações)"],
                align="left"
            ),
            cells=dict(
                values=[ranking["Author"].tolist(), ranking["Impact"].tolist()],
                align="left"
            )
        ),
        row=2, col=2
    )

    fig.update_layout(
        title_text="Dashboard bibliométrico (coautoria, temporalidade, veículos e impacto)",
        height=900,
        margin=dict(l=40, r=40, t=80, b=40)
    )

    return fig


# =========================
# Execução
# =========================

def main():
    df = load_scopus_csv(CSV_PATH)

    # Se seu CSV for “lista de referências”, isso já funciona direto.
    # Se for “lista de artigos analisados” (não referências), então os painéis refletem esse conjunto.

    fig = build_dashboard_figure(df)
    fig.show()


if __name__ == "__main__":
    main()
````
